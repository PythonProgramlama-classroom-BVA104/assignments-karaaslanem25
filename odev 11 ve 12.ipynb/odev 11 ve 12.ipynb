# Araç Kiralama Sistemi
Bu projede abstract class yapısı kullanılarak farklı araç tipleri yönetilmiştir.

```python
# 🚗 FİLO VERİSİ (class YOK)
filo = [
    {"plaka": "34ABC11", "marka": "Fiat", "model": "Egea", "tur": "oto", "musait": True},
    {"plaka": "34DEF22", "marka": "Ford", "model": "Transit", "tur": "ticari", "musait": True},
    {"plaka": "34GHI33", "marka": "Mercedes", "model": "C200", "tur": "lux", "musait": True}
]


# 💰 fiyat hesaplama
def ucret_hesapla(tur):
    if tur == "oto":
        return 1000
    elif tur == "ticari":
        return 1500
    elif tur == "lux":
        return 3000


# 📋 listeleme
def listele():
    for a in filo:
        durum = "Musait" if a["musait"] else "Kirada"
        fiyat = ucret_hesapla(a["tur"])
        print(a["plaka"], a["marka"], a["model"], fiyat, durum)


# 🔑 kiralama
def kirala(plaka):
    for a in filo:
        if a["plaka"] == plaka:
            if a["musait"]:
                a["musait"] = False
                return "Kiralandi"
            else:
                return "Zaten kirada"
    return "Bulunamadi"


# 💰 gelir
def gelir():
    toplam = 0
    for a in filo:
        if not a["musait"]:
            toplam += ucret_hesapla(a["tur"])
    return toplam


# 🎮 MENÜ
while True:
    print("\n1-Liste 2-Kirala 3-Gelir 4-Cikis")
    secim = input()

    if secim == "1":
        listele()

    elif secim == "2":
        p = input("Plaka: ").upper()
        print(kirala(p))

    elif secim == "3":
        print("Gelir:", gelir())

    elif secim == "4":
        break
```
