```python
# Market Stok Takip Programı

# 1. Veri Yapısı: Ürünleri dictionary içinde tutup bir liste içine ekliyoruz
stoklar = [
    {"ad": "Süt", "fiyat": 25.0, "stok": 12},
    {"ad": "Ekmek", "fiyat": 10.0, "stok": 3},
    {"ad": "Yumurta", "fiyat": 3.0, "stok": 30},
    {"ad": "Peynir", "fiyat": 150.0, "stok": 2}
]

# 2. İşlemler:

# En az 3 ürün listeleyelim (for döngüsü ile)
print("--- Mevcut Stok Listesi ---")
for urun in stoklar:
    print(f"Ürün: {urun['ad']} | Fiyat: {urun['fiyat']} TL | Stok: {urun['stok']}")

# Stok sayısı 5'ten az olanları tespit edelim (if koşulu)
print("\n--- Kritik Stok Uyarıları ---")
toplam_deger = 0
for urun in stoklar:
    if urun['stok'] < 5:
        print(f"UYARI: {urun['ad']} kritik seviyede! (Stok: {urun['stok']})")
    
    # Toplam stok değerini hesaplayalım (fiyat * stok)
    toplam_deger += (urun['fiyat'] * urun['stok'])

# Toplam değeri yazdıralım
print(f"\nMarketin toplam stok değeri: {toplam_deger} TL")
```
